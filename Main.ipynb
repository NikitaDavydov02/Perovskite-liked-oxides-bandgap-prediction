## HIll's formula

```python
import re
from collections import Counter
def parse_formula(formula):
    #Extracts element counts from a chemical formula.
    print("Parsing formula: ",formula)
    #pattern = r"([A-Z][a-z]*)(\d*)"
    pattern = r"([A-Z][a-z]*)(\d*\.?\d*)"
    matches = re.findall(pattern, formula)

    element_counts = Counter()
    for element, count in matches:
        element_counts[element] += float(count) if count else 1  # Default to 1 if count is missing

    return element_counts

#converts chemical formula string to Hill's notation
def to_hill_notation(formula):
    #Converts a chemical formula to Hill notation
    element_counts = parse_formula(formula)

    # Hill notation rules
    if "C" in element_counts:
        elements_sorted = ["C", "H"] if "H" in element_counts else ["C"]
        remaining_elements = sorted(e for e in element_counts if e not in ["C", "H"])
        elements_sorted.extend(remaining_elements)
    else:
        elements_sorted = sorted(element_counts.keys())
    hill_formula = " ".join(f"{el}{element_counts[el] if element_counts[el] > 1 else ''}" for el in elements_sorted)
    #print(f':{hill_formula}:')
    return hill_formula
```

## Mulliken electronegativity

```python
from pymatgen.core.periodic_table import Element

#calculate malliken electronegativity of an element
def get_mulliken_en(element_symbol):
    el = Element(element_symbol)
    IE = el.ionization_energies[0]  # First ionization energy in eV
    EA = el.electron_affinity       # Electron affinity in eV

    if IE is None or EA is None:
        return None

    return (IE + EA) / 2

#calculate malliken electronegativity based on Hill formula
def calc_average_electronegativity(formula):
  # Example: For Fe2O3
  comp = Composition(formula)
  # Weighted mean electronegativity
  total_atoms = comp.num_atoms
  mean_en = sum(
    comp[el] / total_atoms * get_mulliken_en(el)
    for el in comp.elements)
  return mean_en
```

```python
df["Average Mulliken electronegativity"] = df["Hill formula"].apply(calc_average_electronegativity)
```

## Valence electron density

```python
#get a dictinoary with indexes of each elements in chemical formla
def split_formula(formula):
  output={}
  elements_with_indexes = formula.split()
  for el in elements_with_indexes:
    #match = re.match(r"([A-Za-z]+)(\d+)$", el)
    match = re.match(r"([A-Za-z]+)(\d+(?:\.\d+)?)$", el)
    if match:
      output[match.group(1)]=float(match.group(2))
    else:
      output[ el]=float(1)
  return output

def get_valence_electrons_number(hill_fomula):
  split = split_formula(hill_fomula)
  print(split)
  v = split.get("O")
  if v == None:
    return 0
  else:
    return 2*v
```

```python
from pymatgen.core.structure import Structure

#calculates unit cell volume from cif file
def getCellVolume(ID):
  print("getCellVolume() ID:",ID)
  if(ID==-1):
    return 0
  file_path=cif_folder+ f"{ID}.cif"
  if os.path.exists(file_path):
    try:
      structure = Structure.from_file(file_path)
    except:
      print('ERROR: Invalid structure for ',ID)
      return 0
  else:
    print('ERROR: CIF file not found for ',ID)
    return 0

  if(structure == None):
    return 0
  return structure.volume
```

```python
#calculates number of formula units Z in a unit cell based on cif file ID
def getCellZValue(ID):
  if(ID==-1):
    return 0
  #file_path=f"/content/drive/MyDrive/University/Artificial intelligence in chemistry/Perovskite project/Perovskite-liked-oxides-bandgap-prediction/Data/CIF/{ID}.cif"
  file_path=cif_folder+ f"{ID}.cif"
  if os.path.exists(file_path):
    try:
      structure = Structure.from_file(file_path)
    except:
      print('ERROR: Invalid structure for ',ID)
      return 0
  else:
    print('ERROR: CIF file not found for ',ID)
    return 0
  if(structure == None):
    return 0
  return structure.composition.num_atoms / structure.composition.reduced_composition.num_atoms
```

```python
#calculates a,b,c unit cell parameters based on cif file ID
def abcExtractionFromMP(ID):
  if(ID==-1):
    return 0
  #file_path=f"/content/drive/MyDrive/University/Artificial intelligence in chemistry/Perovskite project/Perovskite-liked-oxides-bandgap-prediction/Data/CIF/{ID}.cif"
  file_path=cif_folder+ f"{ID}.cif"
  if os.path.exists(file_path):
    try:
      structure = Structure.from_file(file_path)
    except:
      print('ERROR: Invalid structure for ',ID)
      return [0,0,0]
  else:
    print('ERROR: CIF file not found for ',ID)
    return  [0,0,0]

  if(structure == None):
    return  [0,0,0]
  a = structure.lattice.a
  b = structure.lattice.b
  c = structure.lattice.c
  return [a,b,c]
```

```python
from pymatgen.core import Structure
from pymatgen.analysis.local_env import ValenceIonicRadiusEvaluator

#calcualtes packing fraction from CIF file ID
def get_packing_fraction_from_cif(mp_id):
    print(mp_id)
    #cif_folder = "/content/drive/MyDrive/University/Artificial intelligence in chemistry/Perovskite project/Perovskite-liked-oxides-bandgap-prediction/Data/CIF/"
    cif_path = os.path.join(cif_folder, f"{mp_id}.cif")

    if not os.path.isfile(cif_path):
        return np.nan  # Return NaN if file doesn't exist
    try:
      structure = Structure.from_file(cif_path)
    except:
      return 0
    comp = structure.composition

    # Extract radius for each site
    site_radii = []

    try:
      print('Calculating oxidation states')
      structure.add_oxidation_state_by_guess()
    except:
      print('Error')
      return


    for site in structure:
        #el = Element(site.species_string)
        try:
          specie = site.specie
        except AttributeError:
          return 0
        print(site.specie.ionic_radii)
        print(site.specie.oxi_state)
        print('Specie ionic radii ',site.specie.ionic_radii)
        print('Site oxi satte ',site.specie.oxi_state)
        rad =site.specie.ionic_radii.get(site.specie.oxi_state)
        print('got site')
        if(rad is not None):
          site_radii.append(rad)

    # Compute atomic volume per site
    atom_volumes = [(4/3) * np.pi * r**3 for r in site_radii]
    total_atom_volume = sum(atom_volumes)

    # Compute packing fraction
    packing_fraction = total_atom_volume / structure.volume
    print('OK')
    return packing_fraction
    #except:
        #return np.nan  # Return NaN if anything fails (e.g., missing radii)
```

```python
#Average descriptors between Materials Project, Crystallography Open Database, Springer Materials and manual data

def averaging_valence_electron_density(MP,COD,Springer):
  sum=0;
  count=0;
  print(MP)
  print(COD)
  print(Springer)
  if(pd.notna(MP)):
    sum+=MP
    count+=1
  if(pd.notna(COD)):
    sum+=COD
    count+=1
  if(pd.notna(Springer)):
    sum+=Springer
    count+=1
  print('Count: ',count)
  if count==0:
    return 0
  return sum/count


def averaging_4(MP,COD,Springer,manual):
  sum=0;
  count=0;
  print(MP)
  print(COD)
  print(Springer)
  print(manual)
  if(pd.notna(MP)):
    sum+=MP
    count+=1
  if(pd.notna(COD)):
    sum+=COD
    count+=1
  if(pd.notna(Springer)):
    sum+=Springer
    count+=1
  if(pd.notna(manual)):
    sum+=manual
    count+=1
  print('Count: ',count)
  if count==0:
    print("No value")
    return np.nan
  return sum/count
```

```python
def manual_column_outliers_removal(_df, _column):
  num_rows_before = _df.shape[0]

  lower_bound = min_boundries.get(_column)
  upper_bound = max_boundries.get(_column)

  if(lower_bound is None):
    return _df

  df_clean = _df[(_df[_column]>= lower_bound) & (_df[_column] <= upper_bound)]
  num_rows_after = df_clean.shape[0]
  removed = num_rows_before - num_rows_after
  print(f'Column: {_column} removed ouliers: {removed }')
  print(f'min: {lower_bound} max: {upper_bound}')
  return df_clean
```

```python
def IQR_column_outliers_removal(_df, _column):
  num_rows_before = _df.shape[0]
  Q1 = _df[_column].quantile(0.25)
  Q3 = _df[_column].quantile(0.75)
  IQR = Q3 - Q1

  lower_bound = Q1 - 1.5 * IQR
  upper_bound = Q3 + 1.5 * IQR

  df_clean = _df[(_df[_column]>= lower_bound) & (_df[_column] <= upper_bound)]
  num_rows_after = df_clean.shape[0]
  removed = num_rows_before - num_rows_after
  print(f'Column: {_column} removed ouliers: {removed }')
  print(f'min: {lower_bound} max: {upper_bound}')
  return df_clean
```

```python
from scipy.stats import zscore
def Z_score_column_outliers_removal(_df, _column):
  _df['Z-score'] = zscore(_df[_column])
  cleaned_df = _df[_df['Z-score'].abs() <= 3].copy()
  cleaned_df = cleaned_df.drop(columns=['Z-score'])
  return cleaned_df
```

```python
from sklearn.ensemble import IsolationForest

def isolation_forest_outliers_removal(_df, column, contamination_rate=0.05):
    print('Isolation forest')
    iso_forest = IsolationForest(contamination=contamination_rate, random_state=42)
    predictions = iso_forest.fit_predict(_df[[column]])
    
    # 3. Create a temporary column to flag anomalies
    _df['Is_Outlier'] = predictions
    
    cleaned_df = _df[_df['Is_Outlier'] == 1].copy()
    
    cleaned_df = cleaned_df.drop(columns=['Is_Outlier'])
    
    return cleaned_df
```

```python
def proportional_augmentation(_df, x_cols, y_col, alphas):
    sample_frac=0.2

    print('input: ', _df.shape[0])

    if 'is_synthetic' not in _df.columns:
        _df['is_synthetic'] = 0

    mask = (_df[x_cols] != 0)
    valid_rows = _df[mask]
    print('Total available non-zero rows: ', valid_rows.shape[0])

    base_to_augment = valid_rows.sample(frac=sample_frac, random_state=42)
    print(f'Randomly sampled rows for augmentation ({sample_frac*100}%): ', base_to_augment.shape[0])
    #base_to_augment = _df[mask].copy()
    #print('non-zero rows to augment: ', base_to_augment.shape[0])

    _df.loc[base_to_augment.index, 'is_synthetic'] = 1

    #print('non-zero: ', mask.shape[0])
    #to_augment = _df[mask].copy()
    #print('augment: ',  to_augment.shape[0])

    for alpha in alphas:
      to_augment = base_to_augment.copy()
      print('augment: ',  to_augment.shape[0])
      to_augment[x_cols] = to_augment[x_cols].multiply(alpha, axis=0)
      to_augment[y_col] = to_augment[y_col] * alpha
      to_augment['is_synthetic'] = 1 
      _df = pd.concat([_df, to_augment], ignore_index=True)
      print('_df: ',  _df.shape[0])
    
    return _df
```

```python
def proportional_augmentation_with_noise(
    _df, x_cols, y_col, alphas, noise_level=0.05
):
    print("input rows: ", _df.shape[0])

    # Robustness: Handle single string or list of strings
    if isinstance(x_cols, str):
        x_cols = [x_cols]

    # Correct 1D boolean mask filtering
    mask = (_df[x_cols] != 0).all(axis=1)
    base_to_augment = _df[mask].copy()
    print("non-zero rows to augment: ", base_to_augment.shape[0])

    augmented_dfs = [_df]

    for alpha in alphas:
        # 1. Scale the baseline data
        scaled_df = base_to_augment.copy()
        scaled_df[x_cols] = scaled_df[x_cols].multiply(alpha, axis=0)

        # Handle Target Scaling (Linear vs. Logarithmic)
        scaled_df[y_col] = scaled_df[y_col] * alpha

        # 2. Add Relative Gaussian Noise to Features (X)
        for col in x_cols:
            # Standard deviation is a percentage of the scaled value
            std_dev = (scaled_df[col] * noise_level).abs()
            noise = np.random.normal(0, std_dev, size=len(scaled_df))
            scaled_df[col] = scaled_df[col] + noise

        # 3. Add Relative Gaussian Noise to Target (Y)
        std_dev = (scaled_df[y_col] * noise_level).abs()

        noise_y = np.random.normal(0, std_dev, size=len(scaled_df))
        scaled_df[y_col] = scaled_df[y_col] + noise_y

        augmented_dfs.append(scaled_df)

    # Combine efficiently
    final_df = pd.concat(augmented_dfs, ignore_index=True)
    print("final augmented _df rows: ", final_df.shape[0])

    return final_df
```

```python
def draw_actual_predicted(_y_test, _y_pred):
  fig = px.scatter(
    x=_y_test,
    y=_y_pred,
    labels={'x':'Actual', 'y':'Predicted'},
    title='Actual vs Predicted',
    width = 500,
    height = 500,
  )
  if target_col == "Log_rate":
      axis_min, axis_max = -5, 12
      axis_dtick=4
  elif target_col == "Bandgap, eV":
      axis_min, axis_max = 0, 5
      axis_dtick=1
  else:
      # Fallback to data limits if target doesn't match
      axis_min = min(min(_y_test), min(_y_pred))
      axis_max = max(max(_y_test), max(_y_pred))
  fig.update_traces(marker_size=4)
  fig.add_shape(
        type="line",
        x0=axis_min, y0=axis_min,
        x1=axis_max, y1=axis_max,
        line=dict(color="red", dash="dash")
    )
  
  fig.update_xaxes(
        range=[axis_min, axis_max], 
        dtick=axis_dtick, 
        showgrid=True, 
        gridcolor='LightGray'
    )
  fig.update_yaxes(
        range=[axis_min, axis_max], 
        dtick=axis_dtick, 
        showgrid=True, 
        gridcolor='LightGray'
    )
  
  fig.update_xaxes(range=[axis_min, axis_max], showgrid=True, gridcolor='LightGray')
  fig.update_yaxes(range=[axis_min, axis_max], showgrid=True, gridcolor='LightGray')
  fig.update_layout(yaxis=dict(scaleanchor="x", scaleratio=1))

  rmse = np.sqrt(mean_squared_error(_y_test, _y_pred))
  r2 = r2_score(_y_test, _y_pred)

  #fig.add_annotation(
  #  x=0.05, y=0.95, xref="paper", yref="paper",
  #  text=f"R² = {r2:.4f}<br>RMSE = {rmse:.4f}",
  #  showarrow=False,
  #  font=dict(size=14)
  #)
  fig.show()
  return {'r2':r2,'rmse':rmse}
```

```python
from sklearn.ensemble import IsolationForest
from sklearn.neighbors import LocalOutlierFactor
from sklearn.covariance import EllipticEnvelope
from sklearn.preprocessing import RobustScaler

def evaluate_model_with_cv(model, without_additional_scaling = False):
    model_name = type(model).__name__
    print(f"--- Evaluation for {model_name} ---")

    scaler = None
    if(normalization_type == "Stad"):
        print('Std')
        scaler = StandardScaler()
    if(normalization_type == "Norm"):
        print('Norn')
        scaler = MinMaxScaler()
    if(normalization_type == "Robust"):
        print('Robust')
        scaler = RobustScaler()
    else:
        print('No scaler')
    scaled_pipeline = Pipeline([
        ("scaler", scaler),
        ("regressor", model)
        ])
    if(without_additional_scaling):
        scaled_pipeline = Pipeline([("regressor", model)])

    cv = KFold(n_splits=5, shuffle=True)
    custom_splits = []
    if do_augmentation:
        is_synthetic_series = df_aumentation_tracker['is_synthetic']
        is_synthetic_list = is_synthetic_series.tolist()
        #print("Is synthetic list: ", is_synthetic_list)
        #print("Is synthtic len: ", len(is_synthetic_list))
        orig_indices = is_synthetic_series[is_synthetic_series == 0].index.values
        synth_indices = is_synthetic_series[is_synthetic_series == 1].index.values

        print("Orig points", len(orig_indices))   
        print("Synthetic points", len(synth_indices))
                
        for train_orig_idx, test_orig_idx in cv.split(orig_indices):
            print("CV split")
            test_orig = orig_indices[test_orig_idx]
            train_orig = orig_indices[train_orig_idx]
            print("Train len: ", len(train_orig_idx))
            print("Test len: ", len(test_orig_idx))
            
            #print("Train original points: ", train_orig)
            #print("Test original points: ", test_orig)
            #print("synth_points: ", synth_indices)
            train_clean = np.concatenate([train_orig, synth_indices])
            
            custom_splits.append((train_clean, test_orig))
            print("Clean train length:", len(train_clean))
            print("Clean test length:", len(test_orig))
            #print("Train: ", train_clean)
            #print("Test: ", test_orig)
    
    elif remove_outliers != 'None':
        print('Remove outliers')
        orig_indices = X.index.values 
        print('No augmentation')
        for train_orig_idx, test_orig_idx in cv.split(orig_indices):
            print("CV split")
            clean_test_idx = orig_indices[test_orig_idx]
            clean_train_orig = orig_indices[train_orig_idx]
            
            print('Train before outliers: ', len(clean_train_orig))
            print('Test before outliers: ', len(clean_test_idx))
            X_train_fold = X.loc[clean_train_orig].astype(float) 
            #print("Train fold idx len: ", len(clean_train_orig))
            #print("Train fold idx: ", clean_train_orig)

            outlier_preds = []
            if(remove_outliers=="Isolation_forest"):
                print("Isolation forest")
                iso = IsolationForest(random_state=42)
                outlier_preds = iso.fit_predict(X_train_fold)
            elif remove_outliers == "LOF":
                print("LOF")
                # NOTE: For LOF on new training folds, use novelty=True so 
                # fit_predict behaves normally, or just use fit_predict without it.
                lof = LocalOutlierFactor(n_neighbors=20, novelty=True)
                lof.fit(X_train_fold)
                outlier_preds = lof.predict(X_train_fold) # Returns 1 for inlier, -1 for outlier

                # --- METHOD 3: Elliptic Envelope (For normally distributed data) ---
            elif remove_outliers == "EllipticEnvelope":
                print("EllipticEnvelope")
                # contamination specifies the expected proportion of outliers (e.g., 5%)
                ee = EllipticEnvelope(contamination=0.05, random_state=42)
                outlier_preds = ee.fit_predict(X_train_fold)
            elif remove_outliers == "IQR":
                print("IQR")
                q1 = X_train_fold.quantile(0.25,numeric_only=True)
                q3 = X_train_fold.quantile(0.75,numeric_only=True)
                iqr = q3 - q1
                
                lower_bound = q1 - 1.5 * iqr
                upper_bound = q3 + 1.5 * iqr
                
                is_inlier = ((X_train_fold >= lower_bound) & (X_train_fold <= upper_bound)).all(axis=1)
                
                outlier_preds = np.where(is_inlier, 1, -1)
            elif remove_outliers == "Z_Score":
                print("Z_Score")
                # 1. Calculate mean and standard deviation per column from the train fold
                mean = X_train_fold.mean()
                std = X_train_fold.std()
                
                # 2. Compute absolute Z-scores for every cell
                z_scores = np.abs((X_train_fold - mean) / std)
                
                # 3. Find rows where ALL features have a Z-score less than the threshold (Standard is 3.0)
                threshold = 3.0
                is_inlier = (z_scores < threshold).all(axis=1)
                
                # 4. Convert boolean to 1/-1 format
                outlier_preds = np.where(is_inlier, 1, -1)
            elif remove_outliers=="Manual":
                # Start by assuming all rows are inliers (True)
                is_inlier = np.ones(len(X_train_fold), dtype=bool)
                
                for feature, min_ in min_boundries.items():
                    if feature in X_train_fold.columns:
                        low = min_boundries[feature]
                        high = max_boundries[feature]
                        # Create a mask for this specific feature
                        within_bounds = (X_train_fold[feature] >= low) & (X_train_fold[feature] <= high)
                        
                        # Chain it with previous rules (all rules must be True)
                        is_inlier = is_inlier & within_bounds
                    else:
                        print(f"Warning: {feature} not found in training fold columns.")
            
            # Convert boolean True/False to scikit-learn standard 1/-1
            outlier_preds = np.where(is_inlier, 1, -1)
            
            #print("outlier_preds len: ", len(outlier_preds))
            #print("outlier_preds: ", outlier_preds)
            inlier_mask = (outlier_preds == 1)
            #print("inlier_mask len: ",len( inlier_mask))
            #print("inlier_mask: ", inlier_mask)
            clean_train_idx = clean_train_orig[inlier_mask]
            #print('Train idx without outliers len: ', len(clean_train_idx))
            #print('Train idx without outliers: ', clean_train_idx)
            
            custom_splits.append((clean_train_idx, clean_test_idx))
            print("Clean train idx length:", len(clean_train_idx))
            print("Clean test idx length:", len(clean_test_idx))

    scoring_metrics = {
        "mse": "neg_mean_squared_error",
        "r2": "r2",
    }

    cv_results = cross_validate(
        scaled_pipeline,
        X,
        y,
        cv=cv if not do_augmentation else custom_splits,
        scoring=scoring_metrics,
        return_train_score=True,  # Crucial for overfitting diagnostics
    )

    train_rmse_folds = np.sqrt(-cv_results["train_mse"])
    train_r2_folds = cv_results["train_r2"]

    # Validation (Test) metrics
    val_rmse_folds = np.sqrt(-cv_results["test_mse"])
    val_r2_folds = cv_results["test_r2"]

    print(f"Metrics     | Train Average (+/- Std) | Validation Average (+/- Std)")
    print(f"------------|-------------------------|---------------------------")
    print(
        f"RMSE        | {train_rmse_folds.mean():.4f} (+/- {train_rmse_folds.std():.4f}) | {val_rmse_folds.mean():.4f} (+/- {val_rmse_folds.std():.4f})"
    )
    print(
        f"R2 Score    | {train_r2_folds.mean():.4f} (+/- {train_r2_folds.std():.4f}) | {val_r2_folds.mean():.4f} (+/- {val_r2_folds.std():.4f})"
    )
    print("-" * 70)
    print("Test RMSE folds:", val_rmse_folds)
    print("Test R2 folds:", val_r2_folds)
    r2_folds.append(val_r2_folds)
    
    y_pred = cross_val_predict(scaled_pipeline, X,y, cv=cv)
    draw_actual_predicted(y, y_pred)
    res_rows.append([model_name, 
                     train_r2_folds.mean(),val_r2_folds.mean(),train_rmse_folds.mean(),val_rmse_folds.mean(),
                     train_r2_folds.std(),val_r2_folds.std(),train_rmse_folds.std(),val_rmse_folds.std()])
```

```python
def evaluate_model_with_cv_and_outliers_removal(model, without_additional_scaling = False):
    model_name = type(model).__name__
    print(f"--- Evaluation for {model_name} ---")

    scaled_pipeline = Pipeline([
        ("scaler", StandardScaler()), 
        ("regressor", model)
        ])
    if(without_additional_scaling):
        scaled_pipeline = Pipeline([("regressor", model)])

    cv = KFold(n_splits=5, shuffle=True)
    custom_splits = []
    orig_indices = X.index.values 
    if remove_outliers is not 'None':
        for train_orig_idx, test_orig_idx in cv.split(orig_indices):
            print("CV split")
            clean_test_idx = orig_indices[test_orig_idx]
            clean_train_orig = orig_indices[train_orig_idx]
            
            print('Train before outliers: ', len(clean_train_orig))
            X_train_fold = X.loc[clean_train_orig]
            iso = IsolationForest(random_state=42)
            outlier_preds = iso.fit_predict(X_train_fold)
            inlier_mask = (outlier_preds == 1)
            clean_train_idx = clean_train_orig[inlier_mask]
            print("Clean train idx:", clean_train_idx)
            #print('Train after outliers: ', len(clean_train_idx))
            
            custom_splits.append((clean_train_idx, clean_test_idx))
            print("Clean train idx length:", len(clean_train_idx))
            print("Clean test idx length:", len(clean_test_idx))

    scoring_metrics = {
        "mse": "neg_mean_squared_error",
        "r2": "r2",
    }

    cv_results = cross_validate(
        scaled_pipeline,
        X,
        y,
        cv=cv if not do_augmentation else custom_splits,
        scoring=scoring_metrics,
        return_train_score=True,  # Crucial for overfitting diagnostics
    )

    train_rmse_folds = np.sqrt(-cv_results["train_mse"])
    train_r2_folds = cv_results["train_r2"]

    # Validation (Test) metrics
    val_rmse_folds = np.sqrt(-cv_results["test_mse"])
    val_r2_folds = cv_results["test_r2"]

    print(f"Metrics     | Train Average (+/- Std) | Validation Average (+/- Std)")
    print(f"------------|-------------------------|---------------------------")
    print(
        f"RMSE        | {train_rmse_folds.mean():.4f} (+/- {train_rmse_folds.std():.4f}) | {val_rmse_folds.mean():.4f} (+/- {val_rmse_folds.std():.4f})"
    )
    print(
        f"R2 Score    | {train_r2_folds.mean():.4f} (+/- {train_r2_folds.std():.4f}) | {val_r2_folds.mean():.4f} (+/- {val_r2_folds.std():.4f})"
    )
    print("-" * 70)
    print("Test RMSE folds:", val_rmse_folds)
    
    y_pred = cross_val_predict(scaled_pipeline, X,y, cv=cv)
    draw_actual_predicted(y, y_pred)
    res_rows.append([model_name, 
                     train_r2_folds.mean(),val_r2_folds.mean(),train_rmse_folds.mean(),val_rmse_folds.mean(),
                     train_r2_folds.std(),val_r2_folds.std(),train_rmse_folds.std(),val_rmse_folds.std()])
```

```python
def evaluate_and_optimize_model_with_nested_cv(model, grid):
    outer_cv = KFold(n_splits=5, shuffle=True)#random_state=42
    inner_cv = KFold(n_splits=5, shuffle=True)#random_state=42

    results = {}

    # 3. Execute Nested Cross-Validation
    #for model_name, config in models_and_grids.items():
    model_name = type(model).__name__
    print("Model: ",model_name)
    outer_rmse, outer_mae, outer_r2 = [], [], []
    
    # Outer Loop: Evaluate Model Capability
    for train_idx, test_idx in outer_cv.split(X, y):
        print("_Outer loop iteration")
        _X_train, _X_test = X.iloc[train_idx], X.iloc[test_idx]
        _y_train, _y_test = y.iloc[train_idx], y.iloc[test_idx]

        pipeline = Pipeline([
                ('scaler', StandardScaler()),
                ('regressor', model)
            ])
        
        # Inner Loop: Find optimal hyperparameters using GridSearchCV
        # 'neg_mean_squared_error' is used as sklearn maximizes optimization scores
        clf = GridSearchCV(
            estimator=pipeline, 
            param_grid=grid, 
            cv=inner_cv, 
            scoring='neg_mean_squared_error',
            n_jobs=-1
        )
        clf.fit(_X_train, _y_train)
        
        print("Best model is redefined")
        best_model_fold = clf.best_estimator_
        best_parameters_fold = clf.best_params_
        
        predictions = best_model_fold.predict(_X_test)

        fold_rmse = np.sqrt(mean_squared_error(_y_test, predictions))
        fold_r2 = r2_score(_y_test, predictions)
            
        outer_rmse.append(fold_rmse)
        outer_r2.append(fold_r2)
        
            

    mean_rmse, std_rmse = np.mean(outer_rmse), np.std(outer_rmse)
    mean_r2, std_r2 = np.mean(outer_r2), np.std(outer_r2)
    
    print(f"\n Final {model_name} Summary:")
    print(f"  Mean RMSE: {mean_rmse:.4f} (+/- {std_rmse:.4f})")
    print(f"  Mean R2:   {mean_r2:.4f} (+/- {std_r2:.4f})")

    #res_rows.append([model_name, mean_r2,0,mean_rmse,0,std_r2,0,std_rmse,0])

    #y_pred = cross_val_predict(best_model_fold, X,y, cv=outer_cv)
    #draw_actual_predicted(y, y_pred)
    
    evaluate_model_with_cv(best_model_fold, without_additional_scaling=True)

    print("Best params: ", best_parameters_fold)
    best_models[model_name] = best_model_fold
    best_params[model_name] = best_parameters_fold

    return best_model_fold,best_parameters_fold
```
